# **Project Summary -**

EMIPredict-AI is an AI-powered decision-support application designed to evaluate a customer's EMI eligibility and estimate a maximum affordable monthly EMI from a structured financial profile.

The system combines a FastAPI REST backend, XGBoost classification and regression models, SQLAlchemy with SQLite for persistence, MLflow for experiment and model tracking, and a Streamlit presentation layer.

The prediction service performs two complementary tasks. The classification model assigns an EMI outcome among Eligible, High Risk, and Not Eligible while providing class probabilities. The regression model estimates the maximum recommended monthly EMI.

The system also computes the requested monthly EMI from the requested loan amount and tenure, stores customer and prediction records, and returns a structured prediction response with a recommendation.

The final tracked experiments recorded 323,840 training rows and 33 model features. The selected XGBoost classifier achieved 92.03% accuracy and 98.94% ROC-AUC, while the selected XGBoost regression model achieved an R² of 0.9928 and RMSE of 658.73.

# **Problem Statement**


Traditional EMI eligibility decisions may provide only a simple eligible/not-eligible outcome without clearly indicating the applicant's affordability level.

An effective EMI decision-support system should evaluate a structured financial profile and provide:

- EMI eligibility classification
- Class probability information
- Maximum affordable monthly EMI
- Comparison with the requested EMI
- A clear recommendation

EMIPredict-AI addresses this problem using two complementary machine-learning models: an XGBoost classification model for EMI eligibility and an XGBoost regression model for maximum affordable EMI estimation.
